```python
import numpy as np

print("=== TAHAP 1: INPUT DATA MATRIKS ===")

# Ketik matriks koefisien persamaan Anda di sini
A = np.array([[3.0, 2.0, 4.0],
              [2.0, 1.0, 2.0],
              [1.0, 3.0, 5.0]], dtype=float)

# Ketik matriks konstanta hasil di sini
B = np.array([11.0, 6.0, 11.0], dtype=float)

# Perintah untuk menggabungkan kedua matriks
M = np.column_stack((A, B))
n = len(B)

print("Bentuk Matriks Diperbesar [A|B]:")
print(M)
```

```python
print("=== TAHAP 2: PROSES ELIMINASI OBE ===")

# Perulangan untuk memproses diagonal utama matriks
for i in range(n):
    # Perulangan untuk mengubah baris di bawah diagonal menjadi 0
    for k in range(i + 1, n):
        faktor = M[k][i] / M[i][i]
        M[k] = M[k] - faktor * M[i]

print("Matriks Hasil Akhir Segitiga Atas:")
print(M)
```

```python
print("=== TAHAP 3: SUBSTITUSI MUNDUR & OUTPUT ===")

# Wadah kosong penampung hasil variabel x, y, z
X = np.zeros(n)

# Menghitung nilai variabel paling akhir (z)
X[n-1] = M[n-1][n] / M[n-1][n-1]

# Perulangan hitung mundur untuk mendapatkan nilai y dan x
for i in range(n-2, -1, -1):
    X[i] = M[i][n]
    for j in range(i+1, n):
        X[i] = X[i] - M[i][j] * X[j]
    X[i] = X[i] / M[i][i]

# Menampilkan hasil cetak jawaban di layar
print("Solusi Akhir Sistem:")
print(f"Nilai X = {X[0]:.2f}")
print(f"Nilai Y = {X[1]:.2f}")
print(f"Nilai Z = {X[2]:.2f}")
```

```python
import numpy as np  # Mengimpor library NumPy untuk melakukan operasi matriks dan array secara efisien

print("=== TAHAP 1: INPUT DATA MATRIKS ===")

# Mendefinisikan Matriks A (Koefisien angka di depan variabel x, y, dan z)
# Menggunakan tipe data float (desimal) agar hasil pembagian nantinya akurat
A = np.array([[3.0, 2.0, 4.0],
              [2.0, 1.0, 2.0],
              [1.0, 3.0, 5.0]], dtype=float)

# Mendefinisikan Matriks B (Konstanta atau nilai hasil di sebelah kanan tanda sama dengan)
B = np.array([11.0, 6.0, 11.0], dtype=float)

# Menggabungkan Matriks A dan Matriks B secara berdampingan menjadi Augmented Matrix [A|B]
M = np.column_stack((A, B))

# Mengambil jumlah total baris/persamaan berdasarkan panjang ukuran matriks B (nilainya = 3)
n = len(B)

print("Bentuk Matriks Diperbesar [A|B]:")
print(M)  # Menampilkan bentuk augmented matrix awal ke layar


print("=== TAHAP 2: PROSES ELIMINASI OBE ===")

# Perulangan pertama (i) untuk mengontrol baris diagonal utama yang dijadikan acuan (pivot)
for i in range(n):

    # Perulangan kedua (k) untuk memproses semua baris yang terletak di bawah baris acuan (pivot)
    for k in range(i + 1, n):
        # Menghitung faktor pengali: elemen di bawah diagonal dibagi dengan elemen diagonal utamanya
        faktor = M[k][i] / M[i][i]

        # Mengurangi semua elemen di baris ke-k dengan hasil perkalian (faktor * baris acuan ke-i)
        # Langkah ini secara matematis mengubah angka di bawah diagonal utama menjadi 0
        M[k] = M[k] - faktor * M[i]

print("Matriks Hasil Akhir Segitiga Atas:")
print(M)  # Menampilkan matriks segitiga atas yang sudah bersih dari angka di bawah diagonal


print("=== TAHAP 3: SUBSTITUSI MUNDUR & OUTPUT ===")

# Menyiapkan wadah kosong (array berisi angka 0) berukuran n untuk menyimpan solusi nilai X, Y, dan Z
X = np.zeros(n)

# Menghitung nilai variabel paling akhir (Variabel Z / indeks ke-2 pada array)
# Rumus: Konstanta baris terakhir dibagi koefisien Z baris terakhir
X[n-1] = M[n-1][n] / M[n-1][n-1]

# Perulangan mundur dimulai dari indeks baris kedua hingga baris pertama (indeks 0)
for i in range(n-2, -1, -1):
    X[i] = M[i][n]  # Menyimpan nilai konstanta pada baris ke-i terlebih dahulu

    # Mengurangi nilai konstanta dengan perkalian koefisien variabel lain yang sudah ditemukan nilainya
    for j in range(i+1, n):
        X[i] = X[i] - M[i][j] * X[j]

    # Membagi hasil sisa pengurangan dengan nilai koefisien diagonal utama pada baris tersebut
    X[i] = X[i] / M[i][i]

# Menampilkan hasil cetak jawaban akhir sistem ke layar dengan pembulatan 2 angka desimal (.2f)
print("Solusi Akhir Sistem:")
print(f"Nilai X = {X[0]:.2f}")  # Mengambil nilai variabel X pada indeks ke-0
print(f"Nilai Y = {X[1]:.2f}")  # Mengambil nilai variabel Y pada indeks ke-1
print(f"Nilai Z = {X[2]:.2f}")  # Mengambil nilai variabel Z pada indeks ke-2
```
