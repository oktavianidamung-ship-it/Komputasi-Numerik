**Penjelasan Cell:**  
Cell ini berisi identitas mahasiswa serta ringkasan logika Metode Bisection untuk mencari akar persamaan non-linear $f(x) = x^3 - x - 2 = 0$ dengan membagi dua interval secara berulang hingga mencapai batas toleransi galat.

```python
# Penjelasan Cell: Mendefinisikan fungsi persamaan non-linear f(x) serta memasukkan nilai batas bawah (a), batas atas (b), dan batas toleransi error.

# Mendefinisikan fungsi matematika non-linear f(x) = x^3 - x - 2
def fungsi_f(x):
    return x**3 - x - 2  # Mengembalikan hasil perhitungan pangkat tiga dikurangi x dan 2

print("=== TAHAP 1: INPUT PARAMETER AWAL ===")  # Menampilkan judul tahapan pertama ke layar

# Menentukan batas bawah (a) dan batas atas (b) sebagai interval tebakan awal akar
batas_bawah = 1.0  # Titik awal sebelah kiri interval (menggunakan tipe float/desimal)
batas_atas = 2.0   # Titik awal sebelah kanan interval

# Menentukan batas toleransi galat/error sebagai syarat berhentinya perulangan
toleransi_error = 0.0001  # Iterasi berhenti jika selisih (b - a) sudah di bawah angka ini

# Menampilkan nilai parameter awal yang telah dimasukkan ke layar
print(f"Interval Awal : [{batas_bawah}, {batas_atas}]")
print(f"Toleransi     : {toleransi_error}")
```

```python
# Penjelasan Cell: Mengevaluasi nilai fungsi pada kedua batas dan memeriksa syarat Teorema Bolzano f(a) * f(b) < 0.

print("=== TAHAP 2: VALIDASI SYARAT AKAR ===")  # Menampilkan judul tahapan kedua ke layar

# Menghitung nilai fungsi f(x) pada titik batas bawah (a) dan batas atas (b)
fa = fungsi_f(batas_bawah)  # Memasukkan nilai a ke dalam fungsi f(x)
fb = fungsi_f(batas_atas)   # Memasukkan nilai b ke dalam fungsi f(x)

# Menampilkan hasil evaluasi fungsi kedua titik batas dengan pembulatan 4 angka desimal
print(f"Nilai f(a) = {fa:.4f}")
print(f"Nilai f(b) = {fb:.4f}")

# Memeriksa syarat mutlak metode Bisection: hasil kali f(a) dan f(b) harus negatif (< 0)
if fa * fb < 0:
    print("Status: Valid! Interval mengapit akar persamaan.")
else:
    print("Status: Tidak Valid! Ganti tebakan batas awal.")
```

```python
# Penjelasan Cell: Menjalankan perulangan metode bagi dua untuk mempersempit interval hingga menemukan pendekatan akar persamaan.

print("=== TAHAP 3: ITERASI BISECTION & HASIL AKHIR ===")  # Menampilkan judul tahapan ketiga ke layar

iterasi = 1  # Variabel penghitung jumlah langkah perulangan dimulai dari iterasi ke-1
titik_tengah = batas_bawah  # Inisialisasi awal variabel penampung nilai titik tengah (c)

# Mencetak judul kolom (header) agar tampilan progres iterasi berbentuk tabel rapi
print("Iterasi\t\tBatas a\t\tBatas b\t\tTitik c\t\tf(c)")
print("-" * 68)

# Perulangan while terus berjalan selama lebar interval |b - a| masih lebih besar dari toleransi
while abs(batas_atas - batas_bawah) > toleransi_error:
    
    # Rumus inti Bisection: membagi dua interval untuk mendapatkan titik tengah baru (c)
    titik_tengah = (batas_bawah + batas_atas) / 2.0
    
    # Menghitung nilai fungsi pada titik tengah baru f(c)
    fc = fungsi_f(titik_tengah)
    
    # Menampilkan angka batas a, batas b, titik tengah c, dan nilai f(c) pada iterasi saat ini
    print(f"{iterasi}\t\t{batas_bawah:.4f}\t\t{batas_atas:.4f}\t\t{titik_tengah:.4f}\t\t{fc:.4f}")
    
    # Mengecek apakah nilai f(c) tepat sama dengan 0 (akar sejati ditemukan)
    if fc == 0.0:
        break  # Hentikan perulangan paksa karena akar sudah ditemukan secara eksak
        
    # Aturan pergeseran batas interval baru:
    elif fungsi_f(batas_bawah) * fc < 0:
        batas_atas = titik_tengah  # Jika f(a) * f(c) < 0, geser batas atas ke titik c
    else:
        batas_bawah = titik_tengah  # Jika tidak, geser batas bawah ke titik c
        
    iterasi += 1  # Menambah angka penghitung iterasi (+1) untuk putaran berikutnya

# Menampilkan garis penutup tabel dan kesimpulan akhir pendekatan akar persamaan
print("-" * 68)
print(f"Akar hampiran ditemukan pada x = {titik_tengah:.4f} (Iterasi ke-{iterasi-1})")
```
