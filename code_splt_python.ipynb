```python
import numpy as np

# Langkah 1: Input Koefisien Matriks A dan Konstanta Matriks B
# Contoh Sistem Persamaan:
#  2x +  y -  z =  8
# -3x -  y + 2z = -11
# -2x  + y + 2z = -3

print("=== LANGKAH 1: INPUT MATRIKS ===")
A = np.array([[2.0, 1.0, -1.0],
              [-3.0, -1.0, 2.0],
              [-2.0, 1.0, 2.0]])

B = np.array([8.0, -11.0, -3.0])

# Menggabungkan A dan B menjadi Augmented Matrix [A|B]
M = np.column_stack((A, B))
print("Augmented Matrix Awal [A|B]:")
print(M)
```

```python
print("=== LANGKAH 2 & 3: PROSES ELIMINASI GAUSS ===")
n = len(B)

# Proses mengubah matriks menjadi Segitiga Atas
for i in range(n):
    # Cari nilai maksimum di bawah baris saat ini untuk teknik pivoting sederhana
    for k in range(i + 1, n):
        faktor = M[k][i] / M[i][i]
        M[k] = M[k] - faktor * M[i]

print("Matriks Setelah Eliminasi (Matriks Segitiga Atas):")
print(M)
```

```python
print("=== LANGKAH 4 & 5: SUBSTITUSI MUNDUR & OUTPUT ===")
# Membuat array kosong untuk menyimpan nilai x, y, z
X = np.zeros(n)

# Melakukan Substitusi Mundur
X[n-1] = M[n-1][n] / M[n-1][n-1]

for i in range(n-2, -1, -1):
    X[i] = M[i][n]
    for j in range(i+1, n):
        X[i] = X[i] - M[i][j] * X[j]
    X[i] = X[i] / M[i][i]

# Menampilkan Solusi Akhir
print("Himpunan Penyelesaian:")
print(f"Nilai x = {X[0]:.2f}")
print(f"Nilai y = {X[1]:.2f}")
print(f"Nilai z = {X[2]:.2f}")
```
