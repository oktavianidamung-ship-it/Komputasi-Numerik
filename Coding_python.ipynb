```python
import numpy as np

print("=== LANGKAH 1: INPUT DATA MATRIKS ===")

# Mendefinisikan Matriks Koefisien A berdasarkan kasus angka riil nomor 5
A = np.array([[3.0, 2.0, 4.0],
              [2.0, 1.0, 2.0],
              [1.0, 3.0, 5.0]], dtype=float)

# Mendefinisikan Matriks Konstanta Hasil B
B = np.array([11.0, 6.0, 11.0], dtype=float)

# Menggabungkan matriks A dan B menjadi Augmented Matrix
M = np.column_stack((A, B))
n = len(B)

print("Augmented Matrix Awal [A|B]:")
print(M)
```

```python
print("=== LANGKAH 2 & 3: PROSES ELIMINASI GAUSS ===")

# Looping untuk memproses baris diagonal utama
for i in range(n):
    # Proses eliminasi nilai pada baris di bawah diagonal utama
    for k in range(i + 1, n):
        # Menghitung nilai faktor pengali k
        faktor = M[k][i] / M[i][i]

        # Mengurangi baris ke-k dengan (faktor * baris ke-i)
        M[k] = M[k] - faktor * M[i]

print("Matriks Setelah OBE (Menjadi Matriks Segitiga Atas):")
print(M)
```

```python
print("=== LANGKAH 4 & 5: SUBSTITUSI MUNDUR & OUTPUT AKHIR ===")

# Menyiapkan array kosong berukuran n untuk menyimpan jawaban nilai x, y, z
X = np.zeros(n)

# Menghitung nilai variabel terakhir (z) terlebih dahulu
X[n-1] = M[n-1][n] / M[n-1][n-1]

# Melakukan perulangan mundur untuk menghitung nilai y dan x
for i in range(n-2, -1, -1):
    X[i] = M[i][n]
    for j in range(i+1, n):
        X[i] = X[i] - M[i][j] * X[j]
    X[i] = X[i] / M[i][i]

# Menampilkan hasil akhir himpunan penyelesaian variabel
print("Himpunan Penyelesaian Sistem:")
print(f"Nilai Variabel X = {X[0]:.2f}")
print(f"Nilai Variabel Y = {X[1]:.2f}")
print(f"Nilai Variabel Z = {X[2]:.2f}")
```
