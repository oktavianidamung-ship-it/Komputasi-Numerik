```python
# 1. Mendefinisikan fungsi persamaan non-linear f(x) = x^3 - x - 2
def fungsi_f(x):
    return x**3 - x - 2

print("=== TAHAP 1: INPUT PARAMETER AWAL ===")

# 2. Menentukan nilai batas bawah (a), batas atas (b), dan toleransi error
batas_bawah = 1.0
batas_atas = 2.0
toleransi_error = 0.0001

print(f"Interval Awal : [{batas_bawah}, {batas_atas}]")
print(f"Toleransi     : {toleransi_error}")
```

```python
print("=== TAHAP 2: VALIDASI SYARAT AKAR ===")

# 3. Menghitung nilai fungsi pada kedua titik batas
fa = fungsi_f(batas_bawah)
fb = fungsi_f(batas_atas)

print(f"Nilai f(a) = {fa:.4f}")
print(f"Nilai f(b) = {fb:.4f}")

# 4. Mengecek apakah f(a) * f(b) < 0 sebagai syarat mutlak metode Bisection
if fa * fb < 0:
    print("Status: Valid! Interval mengapit akar persamaan.")
else:
    print("Status: Tidak Valid! Ganti tebakan batas awal.")
```

```python
print("=== TAHAP 3: ITERASI BISECTION & HASIL AKHIR ===")

iterasi = 1
titik_tengah = batas_bawah

# 5. Menampilkan header tabel iterasi
print("Iterasi\t\tBatas a\t\tBatas b\t\tTitik c\t\tf(c)")
print("-" * 68)

# 6. Perulangan berjalan selama lebar interval lebih besar dari toleransi
while abs(batas_atas - batas_bawah) > toleransi_error:
    # Hitung titik tengah c
    titik_tengah = (batas_bawah + batas_atas) / 2.0
    fc = fungsi_f(titik_tengah)

    # Cetak progres setiap langkah iterasi
    print(f"{iterasi}\t\t{batas_bawah:.4f}\t\t{batas_atas:.4f}\t\t{titik_tengah:.4f}\t\t{fc:.4f}")

    # Cek jika akar eksak ditemukan
    if fc == 0.0:
        break
    # Tentukan posisi sub-interval baru yang mengapit akar
    elif fungsi_f(batas_bawah) * fc < 0:
        batas_atas = titik_tengah
    else:
        batas_bawah = titik_tengah

    iterasi += 1

# 7. Menampilkan kesimpulan akar persamaan
print("-" * 68)
print(f"Akar hampiran ditemukan pada x = {titik_tengah:.4f} (Iterasi ke-{iterasi-1})")
```
